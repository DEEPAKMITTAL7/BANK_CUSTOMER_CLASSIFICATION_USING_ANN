```python
import tensorflow as tf
import pickle
# load the model
model = tf.keras.models.load_model('bank_customer_churn_model.h5')
```

```python
# load encoder and scaler
with open('encoder.pkl', 'rb') as f:
    le = pickle.load(f)

with open('geo_encoder.pkl', 'rb') as f:
    onehot_encoder = pickle.load(f)

with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40, 
    'Tenure': 3,
    'Balance': 60000, 
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
import pandas as pd
df = pd.DataFrame([input_data])
df.shape
```

```python
df['Gender'] = le.transform(df['Gender'])
```

```python
geo_data = onehot_encoder.transform(df[['Geography']]).toarray()
geo_df = pd.DataFrame(geo_data, columns=onehot_encoder.get_feature_names_out(['Geography']))
geo_df.head()
```

```python
input_data = pd.concat([df.drop('Geography', axis=1), geo_df], axis=1)
input_data_scaled = scaler.transform(input_data)
input_data_scaled.shape
```

```python
#prediction
prediction = model.predict(input_data_scaled)
prediction
```
